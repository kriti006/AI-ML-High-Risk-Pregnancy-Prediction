```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
results = pd.DataFrame({
    'Model':     ['Logistic Regression', 'Decision Tree', 'KNN', 'Random Forest', 'SVM', 'LightGBM','xg boost'],
    'Accuracy':  [0.8218, 0.7953, 0.8306, 0.9043, 0.8999, 0.9146,.9102],
    'ROC-AUC':   [0.8892, 0.7952, 0.9422, 0.9687, 0.9622, 0.9703,.974],
    'Precision': [0.87,   0.78,   0.76,   0.91,   0.90,   0.92,.90],
    'Recall':    [0.76,   0.83,   0.97,   0.90,   0.90,   0.91,.92],
    'F1 Score':  [0.81,   0.80,   0.85,   0.90,   0.90,   0.91,.91]
})

print(results.to_string(index=False))
```

```python
results.set_index('Model').plot(kind='bar', figsize=(12, 6), colormap='Set2')
plt.title('Model Comparison - All Metrics')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
best_model = results.loc[results['Recall'].idxmax(), 'Model']
best_recall = results['Recall'].max()
best_f1_model = results.loc[results['F1 Score'].idxmax(), 'Model']
best_f1 = results['F1 Score'].max()
best_auc_model = results.loc[results['ROC-AUC'].idxmax(), 'Model']
best_auc = results['ROC-AUC'].max()

print("Best Model (Recall)  :", best_model, "->", best_recall)
print("Best Model (F1 Score) :", best_f1_model, "->", best_f1)
print("Best Model (ROC-AUC)  :", best_auc_model, "->", best_auc)
```

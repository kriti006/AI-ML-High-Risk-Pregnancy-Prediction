```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
results = pd.DataFrame({
    'Model':      ['Logistic Regression', 'Decision Tree', 'KNN', 'Random Forest', 'SVM','LightGBM', 'XGBoost', 'CatBoost', 'Gradient Boosting',
                    'Voting Classifier'],
    'Accuracy':   [0.8218, 0.7953, 0.8306, 0.9043, 0.8999,0.9146, 0.9102, 0.9057, 0.8571, 0.9308],
    'ROC-AUC':    [0.8892, 0.7952, 0.9422, 0.9687, 0.9622, 0.9703, 0.9740, 0.9672, 0.92, 0.9797],
    'Precision':  [0.87,   0.78,   0.76,   0.91,   0.90, 0.92,   0.90,   0.9132, 0.8627,0.9246],
    'Recall':     [0.76,   0.83,   0.97,   0.90,   0.90, 0.91,   0.92,   0.8971, 0.8500,0.9382],
    'F1 Score':   [0.81,   0.80,   0.85,   0.90,   0.90,0.91,   0.91,   0.9050, 0.8563,0.9314]
})

print(results.to_string(index=False))
```

```python
results.set_index('Model').plot(kind='bar', figsize=(12, 6), colormap='Set2')
plt.title('Model Comparison - All Metrics')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
best_model = results.loc[results['Recall'].idxmax(), 'Model']
best_recall = results['Recall'].max()
best_f1_model = results.loc[results['F1 Score'].idxmax(), 'Model']
best_f1 = results['F1 Score'].max()
best_auc_model = results.loc[results['ROC-AUC'].idxmax(), 'Model']
best_auc = results['ROC-AUC'].max()

print("Best Model (Recall)  :", best_model, "->", best_recall)
print("Best Model (F1 Score) :", best_f1_model, "->", best_f1)
print("Best Model (ROC-AUC)  :", best_auc_model, "->", best_auc)
```

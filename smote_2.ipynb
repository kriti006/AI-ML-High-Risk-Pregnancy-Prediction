```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
from sklearn.preprocessing import RobustScaler
from imblearn.over_sampling import SMOTE
import joblib
```

```python
df = pd.read_csv('high_risk_pregnancy_realistic_2000 (1).csv')
df.head()
```

```python
df = df.drop(columns=['timestamp', 'device_source'])  # dropping bc smote only understands numeric data
df.head()
```

```python
df = pd.get_dummies(df, columns=['village'], drop_first=True)
df.head()
```

```python
df['edema_severity'] = df['edema_severity'].fillna('None')  # filling nan value as none
df['edema_severity'] = df['edema_severity'].map({'None': 0, 'Mild': 1, 'Moderate': 2, 'Severe': 3})
df['edema_severity']
```

```python
print(df.isnull().sum())
```

```python
df['risk_class'] = df['risk_class'].map({'No Risk': 0, 'Risk': 1})
print(df['risk_class'].value_counts())
```

```python
x = df.drop(columns='risk_class')   # x -> features
y = df['risk_class']                # y -> target
```

```python
patient_ids = x['patient_id']
x = x.drop(columns=['patient_id'])
```

```python
num_cols = ['gestational_age_weeks', 'random_blood_sugar_mg_dL',
            'body_temperature_F', 'hemoglobin_g_dL', 'hba1c_percent',
            'bmi', 'spo2_percent', 'symptoms_score_0_10']

medians_full = {}
for col in num_cols:
    median_val = x[col].median()
    medians_full[col] = median_val
    x[col] = x[col].fillna(median_val)

print("Medians used (full data):", medians_full)
print("\nRemaining NaNs in x:", x.isnull().sum().sum())
```

```python
smote_full = SMOTE(random_state=20)
x_resampled, y_resampled = smote_full.fit_resample(x, y)

print("\nBefore SMOTE (full data):")
print(y.value_counts())
print("\nAfter SMOTE (full data):")
print(y_resampled.value_counts())
```

```python
x_train_pre, x_test_pre, y_train_pre, y_test_pre = train_test_split(
    x_resampled, y_resampled, test_size=0.2, random_state=20, stratify=y_resampled
)

print("\nTrain shape:", x_train_pre.shape)
print("Test shape:", x_test_pre.shape)
print("\nTrain distribution:\n", y_train_pre.value_counts())
print("\nTest distribution:\n", y_test_pre.value_counts())
```

```python
scaler_pre = RobustScaler()
x_train_pre_scaled = scaler_pre.fit_transform(x_train_pre)
x_test_pre_scaled = scaler_pre.transform(x_test_pre)

print("\nScaling done!")
```

```python
joblib.dump(x_train_pre_scaled, 'x_train_presplit_smote.pkl')
joblib.dump(y_train_pre,        'y_train_presplit_smote.pkl')
joblib.dump(x_test_pre_scaled,  'x_test_presplit_smote.pkl')
joblib.dump(y_test_pre,         'y_test_presplit_smote.pkl')
joblib.dump(scaler_pre,         'scaler_presplit_smote.pkl')

print("Saved all pre-split-SMOTE files with new names!")
```

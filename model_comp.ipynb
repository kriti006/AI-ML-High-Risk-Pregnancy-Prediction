```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
results = pd.DataFrame({
    'Model':     ['Random Forest', 'XGBoost', 'Logistic Regression','Decision Tree'],
    'Accuracy':  [0.79,   0.8045, 0.80 ,.67],
    'ROC-AUC':   [0.6664, 0.6848, 0.6848,.5451],
    'Precision': [0.29,   0.39,   0.42,.22],
    'Recall':    [0.16,   0.23,   0.36,.35],
    'F1 Score':  [0.21,   0.29,   0.39 ,.27]
})

print(results.to_string(index=False))
```

```python
results.set_index('Model').plot(kind='bar', figsize=(12, 6), colormap='Set2')
plt.title('Model Comparison - All Metrics')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
best_model = results.loc[results['Recall'].idxmax(), 'Model']
best_recall = results['Recall'].max()
best_f1 = results.loc[results['F1 Score'].idxmax(), 'Model']

print("Best Model (Recall) :", best_model, "→", best_recall)
print("Best Model (F1 Score):", best_f1)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
results = pd.DataFrame({
    'Model':     ['Logistic Regression', 'Random Forest', 'XGBoost', 'Decision Tree', 'SVM', 'KNN','LGBM'],
    'Accuracy':  [0.8045, 0.787,  0.817,  0.6742, 0.7845, 0.6967,.807],
    'ROC-AUC':   [0.6848, 0.6664, 0.6961, 0.5451, 0.6602, 0.591,.6715],
    'Precision': [0.42,   0.29,   0.39,   0.22,   0.32,   0.22,.3721],
    'Recall':    [0.36,   0.16,   0.23,   0.35,   0.22,   0.36,.23],
    'F1 Score':  [0.39,   0.21,   0.29,   0.27,   0.26,   0.27,.28]
})

print(results.to_string(index=False))
```

```python
results.set_index('Model').plot(kind='bar', figsize=(12, 6), colormap='Set2')
plt.title('Model Comparison - All Metrics')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
best_model = results.loc[results['Recall'].idxmax(), 'Model']
best_recall = results['Recall'].max()
best_f1 = results.loc[results['F1 Score'].idxmax(), 'Model']

print("Best Model (Recall) :", best_model, "→", best_recall)
print("Best Model (F1 Score):", best_f1)
```
